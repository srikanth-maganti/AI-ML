## Importing Libraries

```python
import tensorflow as tf
import keras as ke
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
tf.config.list_physical_devices("GPU")
```

# Create a Model

```python
model=tf.keras.Sequential()
model.add(tf.keras.layers.Dense(units=1,input_shape=[1],activation='relu'))

model.summary()
```

```python
model.compile(loss="mean_squared_error",optimizer=tf.keras.optimizers.Adam(1.0))
```

```python
epochs_hist=model.fit(df['Fahrenheit'],df['Celsius'],epochs=100)
```

```python
epochs_hist.history
```

```python
loss=epochs_hist.history['loss']
plt.plot(loss)
```

# Predicting output

```python
temp_f=32
temp_c=model.predict(np.array([temp_f]))
temp_c
```
