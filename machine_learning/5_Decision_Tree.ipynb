### *Importing Libraries*

```python
import pandas as pd
import numpy as np
```

```python
from sklearn.datasets import load_iris
```

### *Load Dataset*

```python
dataset=load_iris()
```

```python
print(dataset.data)
print(dataset.target)
```

### *Summarize Dataset*

```python
# sepel length,sepal width,petal length,petal width
# (Setosa, Versicolour, and Virginica)
```

### *Segregate Dataset into X(Input/IndependentVariable) & Y(Output/DependentVariable)*

```python
x=dataset.data
y=dataset.target
```

### *Splitting Dataset into Train & Test*

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.25,random_state=1)
```

#finding best max_depth value

```python
accuracy=[]
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

for i in range(1,10):
  model=DecisionTreeClassifier(criterion='entropy', max_depth=i,random_state=1)
  model.fit(x_train,y_train)
  pred=model.predict(x_test)
  accuracy.append(accuracy_score(y_test,pred))
plt.figure(figsize=(12,6))
plt.plot(range(1,10),accuracy,linestyle='--' ,marker='o',color='blue')
plt.xlabel("depth value")
plt.ylabel("accuracy")
plt.title("finding best max_depth")
plt.show()
```

### *Training*

```python
from sklearn.tree import DecisionTreeClassifier
model=DecisionTreeClassifier(criterion='entropy',max_depth=3,random_state=1)
model.fit(x_train,y_train)
```

### *Prediction for all Test Data*

```python
data=[[7,1,4,2]]
res=model.predict(data)
if(res==0):
  print("iris setosa")
elif res==1:
  print("vericolor")
else:
  print("verginica")
```

```python
y_pred=model.predict(x_test)
```

### *Evaluating Model*

```python

from sklearn.metrics import accuracy_score
print(accuracy_score(y_test,y_pred))
```
