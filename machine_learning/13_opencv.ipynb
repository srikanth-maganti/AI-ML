## Continues image captureing until press a key


```python
import time
```

```python
count=0
image=cv2.VideoCapture(0)
while True:
    _,frame=image.read()
    cv2.imshow("frame",frame)
    if cv2.waitKey(1)==ord('q'):
        break
    print(frame.shape)
    cv2.imwrite("frme"+str(count)+".jpg",frame)
    count=count+1
    time.sleep(2)
image.release()
cv2.destroyAllWindows()
```

## to capture video

```python
image=cv2.VideoCapture(0)
fourcc_code=cv2.VideoWriter_fourcc(*'XVID')
video=cv2.VideoWriter("vid.avi",fourcc_code,20.0,(640,480))

while True:
    _,frame=image.read()
    video.write(frame)
    cv2.imshow("frame",frame)
    if(cv2.waitKey(100)==ord('q')):
        break

video.release()
image.release()
cv2.destroyAllWindows()
```

# Moving Object Detection

```python
import cv2
import imutils
from datetime import datetime
import time
```

```python
cam=cv2.VideoCapture(0)
time.sleep(1)# to stabilize cam
first_frame=None
area=1000

while True:
    _,img=cam.read(1)
    text="normal"
    img=imutils.resize(img,width=500) #resizeing for easy computation
    grey=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
    blur=cv2.GaussianBlur(grey,(21,21),0)
    if first_frame is None:
        first_frame=blur
        continue
    img_diff=cv2.absdiff(first_frame,blur)#compare two images highlight changed
    threshimg=cv2.threshold(img_diff,25,255, cv2.THRESH_BINARY)[1] # map white and black to motion and non motion
    threshimg = cv2.dilate(threshimg, None, iterations=2) #fill spaces make it continures

    cnts = cv2.findContours(threshimg.copy(), cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    cnts = imutils.grab_contours(cnts)

    for c in cnts:
        if cv2.contourArea(c)<area:
            continue
        (x,y,w,h)=cv2.boundingRect(c)
        cv2.rectangle(img,(x,y),(x+w,y+h),(0,255,0),2)
        text="moving object detected"
    if text!="Normal":
        now=datetime.now()
        current=now.strftime("%H:%M:%S")
        print(text,current)
    cv2.putText(img, text, (10,20), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0,0, 255), 2)
    cv2.imshow('camfeed', img)
    if cv2.waitKey(1)==ord('q'):
        break
cam.release()
cv2.destroyAllWindows()



```
