# Creating functional model for multiple inputs

```python
"""
lets think of this problem ,how can predict price of product with given image and text description,by predicting on individual input and then average or obtainging feature vectors from each type of input and then concatenating them and feed to dense layers to predict price

"""
```

```python
input1=Input(shape=(2,))
input2=Input(shape=(3,))
```

```python
x=Dense(224,activation="relu")(input1)
x1=Dense(128,activation="relu")(x)
x2=Dense(64,activation="relu")(x1)
```

```python
y=Dense(512,activation="relu")(input2)
y1=Dense(224,activation="relu")(y)
y2=Dense(128,activation="relu")(y1)
y3=Dense(64,activation="relu")(y2)

```

```python
combined=concatenate([x2,y3])
z=Dense(32,activation="relu")(combined)
z1=Dense(1,activation="linear")(z)
```

```python
model=Model(inputs=[input1,input2],outputs=z1)
```

```python
from keras.utils import plot_model
plot_model(model)
```
