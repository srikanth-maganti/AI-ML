# Parts of Speech

```python
nlp=spacy.load('en_core_web_sm')
nlp
```

```python
document=nlp('ing Natural Language Processing. The course name is Diploma in Data Science')
```

```python
for token in document:
  print(token.text,token.pos_)
```

# Lemmatization and Stemming

Lemmatization: meaning of the word based on the dictionary (morphological analysis) - extract the base word


```python
for token in document:
  print(token.text,token.lemma_)
```

Stemming: extract the root of the word

```python
import nltk
```

```python
stemmer=nltk.stem.PorterStemmer()
stemmer.stem('learning')
```

```python
for token in document:
  print(token.text,token.lemma_,stemmer.stem(token.text))
```

# Named entity recognition(NER)

```python
text = 'IBM is a US company and located in Chennai, started in the year 1990. this year revence is 300 billion dolars'
```

```python
document=nlp(text)
```

```python
for entity in document.ents:
  print(entity.text,entity.label)
```

```python
from spacy import displacy
displacy.render(document,style='ent')
```

# Stop words

```python
from spacy.lang.en.stop_words import STOP_WORDS
print(STOP_WORDS)
```

```python
#using nlp method
nlp.vocab['chennai'].is_stop
```

```python
for token in document:
  if nlp.vocab[token.text].is_stop:
    print(token.text)
```

```python
for token in document:
  if not nlp.vocab[token.text].is_stop:
    print(token.text)
```
